# Partie B – Normalisation Min-Max 


1. Sélectionner les variables Age, Revenu_annuel_DH, Solde_DH et Anciennete_mois. 


```python
variables_num = ['Age', 'Revenu_annuel_DH', 'Solde_DH', 'Anciennete_mois']
df_num = df[variables_num].copy()
```

2. Vérifier et traiter les valeurs manquantes de ces variables avant d’appliquer le scaler. Indiquer la méthode 
retenue et la justifier. 

```python
df_num.isna().sum()
```

Méthode retenue et justification :Méthode : Imputation par la médiane de chaque colonne.   Justification : La médiane est une mesure robuste insensible aux valeurs extrêmes (outliers) et aux distributions asymétriques, contrairement à la moyenne qui peut être fortement biaisée.

```python
for col in variables_num:
    df_num[col] = df_num[col].fillna(df_num[col].median())
```

3 . Appliquer MinMaxScaler aux variables numériques sélectionnées. 



```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_minmax = pd.DataFrame(scaler.fit_transform(df_num), columns=variables_num)
# df_minmax = scaler.fit_transform(df_num)
```

- Sur les unités de mesure : Supprime toutes les unités (DH, années, mois) pour ramener toutes les caractéristiques sur une même échelle sans unité $[0, 1]$, empêchant une variable ayant de grands chiffres (ex: Revenu_annuel_DH) de dominer le modèle.

# Partie E – Encodage des variables ordinales 


1 . Définir les catégories dans le bon ordre avec OrdinalEncoder. 


```python
from sklearn.preprocessing import OrdinalEncoder

# Traitement préalable des valeurs manquantes par le mode
df['Niveau_etudes'] = df['Niveau_etudes'].fillna(df['Niveau_etudes'].mode()[0])
df['Satisfaction'] = df['Satisfaction'].fillna(df['Satisfaction'].mode()[0])

# Ordres respectifs
ordre_etudes = ['Bac', 'Licence', 'Master', 'Doctorat']
ordre_satisfaction = ['Faible', 'Moyenne', 'Élevée', 'Très élevée']

ord_enc = OrdinalEncoder(categories=[ordre_etudes, ordre_satisfaction])
```

2 . Transformer Niveau_etudes et Satisfaction en valeurs numériques. 


```python
matrice_ord = ord_enc.fit_transform(df[['Niveau_etudes', 'Satisfaction']])
df_ord = pd.DataFrame(matrice_ord, columns=['Niveau_etudes_encoded', 'Satisfaction_encoded'])
```

```python
df_ord['Satisfaction_encoded'].unique()
```

3 . Afficher les valeurs originales et encodées pour quelques lignes. 


```python
comparaison = pd.concat([df[['Niveau_etudes', 'Satisfaction']], df_ord], axis=1)
print(comparaison.head())
```

4 . Expliquer la différence entre un encodage nominal et un encodage ordinal. 


- Encodage Nominal (One-Hot Encoding) : Utilisé pour des variables sans ordre hiérarchique (ex: Ville). Il crée $N$ colonnes binaires ($0/1$) afin de ne pas introduire de relation mathématique d'ordre inexistante entre les catégories.  
- Encodage Ordinal : Utilisé quand un ordre ou un rang logique existe (ex: Niveau d'études). Il convertit les catégories en un seul vecteur d'entiers séquentiels ($0, 1, 2, 3...$), préservant ainsi la hiérarchie tout en conservant une seule colonne.
